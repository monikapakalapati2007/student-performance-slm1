```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

from sklearn.model_selection import train_test_split

# First split: 80% training and 20% temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# Second split: Divide temporary data equally
validation_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

# Display the dataset sizes
print("Training records:", len(train_df))
print("Validation records:", len(validation_df))
print("Testing records:", len(test_df))
```

```python

import os

# Create folder
os.makedirs("student_performance_parquet", exist_ok=True)

# Save files in Parquet format
train_df.to_parquet(
    "student_performance_parquet/train.parquet",
    index=False
)

validation_df.to_parquet(
    "student_performance_parquet/validation.parquet",
    index=False
)

test_df.to_parquet(
    "student_performance_parquet/test.parquet",
    index=False
)

print("All three Parquet files created successfully!")
```
